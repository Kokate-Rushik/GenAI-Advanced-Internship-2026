### 1. Employee Performance Bonus Eligibility

**Problem Statement**: A company evaluates employees performance score at the end of the year. You are given a dictionary containing employee names and their performance score.

**Requirements:**

- Identify the highest performance score.
- Handle ties if multiple employees have the same highest score.
- Display all employees eligible for the top performance bonus.


```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

max_per = max(employees.values())
res = ""
for key, val in employees.items():
    if val == max_per:
        res += f"{key}, "

print(f"Top Performers Eligible for Bonus: {res[:-2]} (Score: {max_per})")
```

### 2. Search Query Keyword Analysis

**Problem Statement**: An e-commerce website stores customer search queries. You are given a search query sentence entered by a user.

**Requirements**:
- Convert the input to lowercase.
- Ignore common punctuation.
- Count the frequency of each keyword.
- Display only keywords searched more than once.


```python
user_query = "Buy mobile phone buy phone online"

res = {}

for i in user_query.split():
    res[i.lower()] = res.get(i.lower(), 0) + 1

res = {k: v for k, v in res.items() if v!=1}

print(res)
```

### 3 Sensor Data Validation

**Problem Statement**:
A factory collects sensor readings every hour. Each reading is stored in a list where the index represents the hour and the value represents the sensor reading.

**Requirements**:
- Identify readings that are even numbers (valid readings).
- Store them as (hour_index, reading_value) pairs.
- Ignore odd readings (invalid readings).


```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

res = []

for i in range(len(sensor_readings)):
    if sensor_readings[i] %2 == 0:
        res.append((i, sensor_readings[i]))
    else: continue

print(f"Valid Sensor Readings (Hour, Value):\n{res}")
```

### 4. Email Domain Usage Analysis

**Problem Statment**:
A company wants to analyze which email providers its users are using. You are given a list of employee email IDs.

**Requirements**:
- Count how many users belong to each email domain.
- Calculate the percentage usage of each domain.


```python
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
n = len(emails)
res = {}

for i in emails:
   res[i.split('@')[1]] = res.get(i.split('@')[1], 0) + 1

for k, v in res.items():
    print(f"{k}: {(v/n)*100}")
```

### 5. Sales Spike Detection

**Problem Statment**: A retail company tracks daily sales. Sudden spikes in sales may indicate promotions or unusual activity.

**Requirements**:
- Calculate the average daily sales.
- Detect days where sales are more than 30% above average.
- Display the day number and sale value.


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

aver = sum(sales)/6
threshold_value =aver + aver*0.3

print(f"Average sales per day is: {aver}")
print(f"30% above average sales per day is: {threshold_value}\n")

for i in range(len(sales)):
    if sales[i] > threshold_value:
        print(f"Day {i+1}: {sales[i]}")

```

### 6. Duplicate User ID Detection

**Problem Statment**: A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.
