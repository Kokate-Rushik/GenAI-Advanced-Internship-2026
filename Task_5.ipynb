# 1 Social Media – Post Engagement Analyzer

- Otherwise → Normal Engagement


```python
'''
For this question i will assume we pass dict of likes from various social media as input
'''

def cal_total_likes(likes):
    '''
    total_likes = 0
    for v in likes.values():
        total_likes += v
    return total_likes
    '''

    return sum(likes.values())

user_inp = [int(x) for x in input("Enter likes from social media (in order Insta, X, YT, FB, others):- ").split()]
user_inp = (user_inp + [0]*5)[:5]
social_media_likes = {
    "Instagram": user_inp[0],
    "Twitter": user_inp[1],
    "YouTube": user_inp[2],
    "Facebook": user_inp[3],
    "Others": user_inp[4]
}

total_likes = cal_total_likes(social_media_likes)
print(social_media_likes)
print(f"Total likes:- {total_likes}")
print("Post status:- ", end='')
print("Viral Post") if total_likes >= 1000 else print("Normal Engagement")
```

#  2. Healthcare – Medicine Stock Alert System

**Problem Statement**:- Create a function to monitor medicine stock levels.
**Rules**:-
- Stock < 10 → Low Stock Alert
- Stock ≥ 10 → Stock Sufficient


```python
def monitor_stock_lvl(stock_qty):
    for i in stock_qty:
        status = "Low Stock Alert" if i[1] < 10 else "Stock Sufficient"
        print(f"{i[0]} → {status}")
    
medicine_stock_list = [['Paracetamol', 8], ['Amoxicillin', 25], ['Ibuprofen', 5], ['Cetirizine', 12], ['Metformin', 9]]
monitor_stock_lvl(medicine_stock_list)
```

# 3: Agriculture – Rainfall Adequacy Checker

**Problem Statement**:-  Create a function that analyzes seasonal rainfall data.

**Rules**:-
- Calculate average rainfall
- Average ≥ required level → Adequate Rainfall
- Otherwise → Inadequate Rainfall

```python

def analyze_rainfall_data(data, req_lvl):
    av_rainfall = sum(data)/len(data)
    status = "Adequate Rainfall" if av_rainfall >= req_lvl else "Inadequate Rainfall"
    print(f"Average Rainfall:- {av_rainfall}")
    print(f"Rainfall Status:- {status}")
    

# values in mm
season_rainfall_data = [730, 492, 30, 120] # (in order of Monsoon, Autumn, Winter, Summer)
analyze_rainfall_data(season_rainfall_data, 200) # assuming required rainfall is 200mm
```

# 4. Social Media – Duplicate Account Detection

**Problem Statement**:- Create a function to detect duplicate usernames in a system.

**Rules**:- 
- If duplicates exist → Flag accounts
- Else → Accounts are unique

```python

def detect_dup_users(data):
    dup = []
    unique = []
    for i in data:
        if i not in unique:
            unique.append(i)
        else:
            dup.append(i)
    
    if dup:
        print("Duplicate Accounts found:- Yes")
        print(dup)
    else:
        print("Duplicate Accounts found:- No")

def detect_dup_users_with_sets(data):
    unique_users = set(data)
    if len(unique_users) != len(data):
        print("Duplicate Accounts found via set operations:- Yes")
    else:
        print("Duplicate Accounts found via set operations:- No")

usernames_list = ["arjun_kumar", "priya_sharma", "rahul_v", "sneha_patil", "arjun_kumar", "anita_desai", "rahul_v", "amit_shah", "sneha_patil"]
detect_dup_users(usernames_list)
print()
detect_dup_users_with_sets(usernames_list)
```

# 6. Agriculture – Premium Crop Price Filter

**Problem Statement**:- Create a function that filters crop prices above a premium threshold.

**Rules**:- 
- Price > ₹2000 → Premium Crop
- Else → Regular Crop

```python
import random

def filter_premium_crop(data):
    premium_crop = []
    for i in data:
        if i[1] >= 2000:
            premium_crop.append(i[1])
    print(f"Premium Crops:- {premium_crop}")

crops = ['Rice', 'Maize', 'Chickpea', 'Kidneybeans', 'Pigeonpeas', 'Mothbeans', 'Mungbean', 'Blackgram', 'Lentil', 'Pomegranate']
crop_price = []

for i in crops:
    price = random.randint(1,100)*50
    crop_price.append([i, price])

filter_premium_crop(crop_price)
```

# 7. System Monitoring – Application Health Checker

**Problem Statement**:- Create a function to evaluate application health based on error count.

**Rules**:-
- Errors = 0 → Healthy
- Errors ≤ 5 → Minor Issues
- Errors > 5 → Critical Issues

```python
import random

def eval_health(err_count):
    if err_count < 0: print("Error")
    status = ""
    if err_count == 0:
        status = "Healthy"
    elif err_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print(f"Error Count:- {err_count}")
    print(f"System Status:- {status}")

for i in range(5):
    health_error_count = random.randint(0,10)
    eval_health(health_error_count)
```

# 8. Banking – Daily Transaction Limit Checker

**Problem Statement**:- Create a function to validate daily transaction amount.

**Rules**:-
- Daily limit = ₹50,000
- Amount within limit → Approved
- Amount exceeds limit → Rejected

```python
todays_transaction_amount = 0

def validate_transaction(amt):
    global todays_transaction_amount
    print(f"Transaction Amount:- {amt}")
    if todays_transaction_amount+amt >= 50000:
        print("Amount exceeds limits → Reject\n")
    else:
        print("Amount within limit → Accepted\n")
        todays_transaction_amount += amt

for i in range(10):
    amt = random.randint(200,20000)
    validate_transaction(amt)
    if i == 5:
        todays_transaction_amount = 0
        print("\nNext Day\nDaily Limit Reset to Zero\n")

print(f"Daily Limit Status:- {todays_transaction_amount}")
```

# 10. Smart Electricity Bill Analyzer

**Problem Statement**:-  You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.

**Billing Rules**:-
- First 100 units → ₹3 per unit
- Next 100 units (101–200) → ₹5 per unit
- Above 200 units → ₹7 per unit

**Usage Classification**:-
+ Total bill < ₹500 → Low Usage
+ ₹500 – ₹1500 → Moderate Usage
+ ₹1500 → High Usage

```python
def analyze_electric_bill(consumption_list):
    total_units_consumed = sum(device[1] for device in consumption_list)
    remaining_units = total_units_consumed

    first_checkpoint = min(remaining_units, 100)
    remaining_units -= first_checkpoint
    bill_amount = (first_checkpoint*3)
    if remaining_units > 0:
        second_checkpoint = min(remaining_units, 100)
        remaining_units -= second_checkpoint
        bill_amount +=(second_checkpoint*5) 
    if remaining_units > 0:
        third_checkpoint = remaining_units
        bill_amount = (third_checkpoint*7)
    
    if bill_amount<500:
        status="Low Usage"
    elif 500 <= bill_amount <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    print(f"Total units consumed:- {total_units_consumed}")
    print(f"Total bill:- {bill_amount}")
    print(f"Usage Classification:- {status}")


device_usage_data = [["LED Bulbs", 30],   ["Ceiling Fans", 80], ["Air Conditioner", 120], ["Refrigerator", 90], ["Water Heater", 110]]
analyze_electric_bill(device_usage_data)
```
