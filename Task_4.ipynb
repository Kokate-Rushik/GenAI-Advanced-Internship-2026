## 1. **Mobile Recharge Validation System**

**Problem Statement**: Create a function to validate a mobile recharge request.

**Rules**:
- Recharge amount must be ≥ ₹50
- Valid plans: 199, 299, 399, 599
- User should be allowed to retry if invalid

```python
class InvalidPlan(Exception): pass

valid_plans = [199, 299, 399, 599]
print(f"Valid Plans: {valid_plans}")

while True:
    recharge_amt = 0
    try:
        recharge_amt = int(input("Enter Recharge amount:- "))
        if recharge_amt not in valid_plans:
            raise InvalidPlan(f"Select suitable valid plan! Recharge amount {recharge_amt}")
        break
    except ValueError:
        print("ValueError:- Please enter recharge amount in numbers")
    except InvalidPlan as e:
        print(f"Invalid Plan Error:- {e}")

print(f"Recharge amount:- {recharge_amt}")
```

## 3. **Student Result Processing System**

**Problem Statement**: Create a function that:
- Accepts a list of student marks
- Calculates average marks
- Prints Pass if average ≥ 50 else Fail

```python
def eval_st(marks):
    average = sum(marks)/len(marks)
    result = "passed" if average >= 50 else "failed"
    print(f"Student {result}. Average marks:- {average}")

student_marks = [int(x) for x in input("Enter marks of 5 subject, separated by white space:- ").split()]
student_marks = (student_marks + [0]*5 )[:5]
print(f"Student marks list:- {student_marks}")
eval_st(student_marks)
```

## 4. **Cab Fare Estimator with Retry Option**

**Problem Statement**: Create a cab fare calculator:
- Base fare = ₹50
- ₹12 per km
- Peak hour → 25% extra
- Allow retry using while loop


```python
import random # to simulate isPeakHour?

def calculateFare(km):
    base_fare = 50
    rate_km = 12

    fare = base_fare + (km * rate_km)

    isPeakHour = random.randint(0,1)

    fare = fare + fare*.25 if isPeakHour == 1 else fare
    if isPeakHour==1: print("\nPeak hour!")
    return fare

retry = "yes"
while retry == "yes":
    try:
        distance_input = int(input("Enter distance in km:- "))
    except ValueError:
        print("ValueError:- Enter distance in number")
        continue
    fare = calculateFare(distance_input)
    print(f"Fare:- ₹{fare} for distance {distance_input}km")
    retry = input("\nWant to retry fare calculation? (Yes/No):- ").lower()
    retry = "yes" if retry in ['yes', 'y', '1'] else "no"
```

## 6. **Password Strength Checker**

**Problem Statement**: Create a function to check password strength.

**Rules**:
- Minimum length = 8
- Must contain at least one digit
- Must contain at least one special character (`@`, `#`, `$`)

```python
def passwd_strength(passwd):
    special = "@#$%^&!"
    n = len(passwd)
    has_digit = any(char.isdigit() for char in passwd)
    has_special_char = any(char in special for char in passwd)
    if n>=8:
        if has_digit:
            if has_special_char:
                print(f"Password '{passwd}' is powerfull!")
            else:
                print(f"Password '{passwd}' missing special characters!")
        else:
            print(f"Password '{passwd}' missing digits!")
    else:
        print(f"Password '{passwd}' length must greater than 8!")

for i in range(4):
    passwd_input = input("Enter password to check it's strength:- ")
    passwd_strength(passwd_input)
    print()
```
