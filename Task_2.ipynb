1. Unique Word in a Sequence

   Problem Statement:- Given a sentence, identify all unique words and display:
   - The count of unique words
   - The unique words themselves


```python

user_input = input("Enter Sentence:- ")
unique_words = set(user_input.split())
print(f"Original sentence:- {user_input}")
print(f"Unique words count:- {len(unique_words)}")
print(f"Unique words:- {unique_words}")
```

2. Highest Salary from the Employee Data

    Problem Statement:- Store employee names and their salaries, and determine which employee has the highest salary

```python

employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}
highest_salary = 0
res = ''
for key, value in employees.items():
    if employees[key] > highest_salary:
        res = key
        highest_salary = employees[key]

if res:
    print(f"Employee with highest salarry is:- {res}")
```

3. Find Maximum and Minimum Values

4. Count Products Above a Price Threshold

    Problem Statement:- Count how many products have a price greater than 1000.

```python
price = [450, 1200, 899, 1500, 300]

# if price has to be user_inputs then:-
# price = [ int(x) for x in input("Enter prices separated by space:- ").split()]

count = sum(1 for i in price if i > 1000)
print(f"Products above 1000:- {count}")
```

5. Calculate Attendance Percentage

    Problem Statement:- Calculate the attendance percentage from a list containing "P" (Present) and "A" (Absent).

```python
attendance = ['P','P','A','P','P']
percent_percent = (sum(1 for i in attendance if i == 'P')/len(attendance))*100
print(f"Attendance Percentage: {percent_percent:.1f}%")
```

7. Count Character Frequency

    Problem Statement:- Count how many times each character appears in a given string.

    I will assume that a character in uppercase and lowercase is count as same character. i.e. 'C' = 'c'. Also i will ignore white spaces.

```python
input_str = input("Enter a string")

res = dict()
for i in input_str:
    if i == ' ': continue
    i.lower()
    res[i.lower()] = res.get(i.lower(), 0) + 1

print(res)
```

8. Convert List to Tuple

10. Calculate Average Marks

    Problem Statement:- Calculate the average marks of a student from a list of marks.

```python
st_marks = [ int(x) for x in input("Enter an marks separated by space:- ").split()]

average = sum(st_marks)/len(st_marks)
print(f"Marks list:- {st_marks}")
print(f"Average Marks:- {average}")
```
