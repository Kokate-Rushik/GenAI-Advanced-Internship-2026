1. User Login Check

    Problem Statement: Given a username and password, check whether login is successful.
    username = "admin"
    password = "1234"

    assuming that username = "admin" (string) and password = "1234" (string)

```python
username = input("Enter username:- ")
password = input("Enter password:- ")


if username == "admin" and password == "1234":
    print("Login Successful!")
else:
    print("Invalid Credentials")
```

2. Pass / Fail Analyzer

    Problem Statement:- Analyze student result.
    <br>marks = [45, 78, 90, 33, 60]

```python
marks = [45, 78, 90, 33, 60]

passed = sum(1 for i in marks if i >= 50)

print(f"Count of student passed:- {passed}")
print(f"Count of student failed:- {len(marks)-passed}")
```

3. Simple Data Cleaner

    Problem Statement:- Clean and standardize user names.
    <br>names = [" Alice ", "bob", " CHARLIE "]

    **Requirements**:
    - Remove extra spaces
    - Convert all names to lowercase
    - Print the cleaned list


```python
names = [" Alice ", "bob", " CHARLIE "]
cleaned = [i.strip().lower() for i in names]
print(f"Cleaned List of names:- {cleaned}")
```

4. Message Length Analyzer

    Problem Statement:- Analyze message sizes.
    <br>messages = ["Hi", "Welcome to the platform", "OK"]

    **Requirements**:
    - Print the length of each message
    - Flag messages longer than 10 characters


```python
msg = ["Hi", "Welcome to the platform", "OK"]
flagged = []

for i in range(len(msg)):
    print(f"Message:- '{msg[i]}', Length :- {len(msg[i])}\n")
    if len(msg[i]) > 10:
        flagged.append(i)

print("Messages that are flagged")
for i in flagged:
    print(f"'{msg[i]}'")
```

5. Error Message Detector

    Problem Statement:- Detect error messages from system logs.
    <br>logs = ["INFO", "ERROR", "WARNING", "ERROR"]

    **Requirements**:
    - Count the number of "ERROR" entries
    - Print the total error count


```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

err_count = sum(1 for i in logs if i == "ERROR")

print(f"Total count of error message is:- {err_count}")
```
