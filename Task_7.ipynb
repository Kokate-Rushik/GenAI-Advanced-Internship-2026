## 1. Smart Light Controller 

**Problem Statement**: Create a class to control a smart light that can be turned ON or OFF and display its status.

**Sample Input**:<br>
- Light Name: Bedroom Light 
- Action: ON <br>

**Expected Output**:<br>
Bedroom Light is ON

```python
class Light_Bulb:
    def __init__(self, name, state=False):
        self.light_on = state
        self.name = name
    
    def turn_on_light(self):
        self.light_on = True
    
    def turn_off_light(self):
        self.light_on = False

    def status(self):
        return "ON" if self.light_on==True else "OFF"

    def __str__(self):
        return f"Light Name: {self.name} Light\nAction: {self.status()}"

# default during object initialization
bedroom_light = Light_Bulb("Bedroom")
print(bedroom_light)

# Turning on light with function
bedroom_light.turn_on_light()
print(bedroom_light)

# Turning off light with function
bedroom_light.turn_off_light()
print(bedroom_light)
```

## 2. Employee ID Card System 
**Problem Statement**: Create a class that stores employee name, ID, and department and displays ID card details.

**Sample Input**:<br>
Employee Name: Rahul <br>
Employee ID: EMP102 <br>
Department: IT <br>
 
 
**Expected Output**<br>
Employee ID Card <br>
Name: Rahul <br>
ID: EMP102 <br>
Department: IT<br>

```python
class Emp:
    def __init__(self, eid, name, dept):
        self.eid = eid
        self.name = name
        self.dept = dept 
    
    def __str__(self):
        return f"Employee ID Card\nName: {self.name}\nID: EMP{self.eid}\nDepartment: {self.dept}\n"

emp1 = Emp(101, "Rishi", "IT")
emp2 = Emp(102, "Rahul", "IT")
print(emp1)
print(emp2)
```

## 3. Mobile Contact Record 

**Problem Statement**: Create a class to store contact name and phone number and display contact information. 

**Sample Input**:<br>
Contact Name: Anita <br>
Phone Number: 9876543210 <br>

**Expected Output**<br>
Contact Saved <br>
Name: Anita <br>
Phone: 9876543210<br>

```python
class Contact:
    def __init__(self, name, phone_no):
        self.name = name
        self.phone_no = phone_no
        print("Contact Saved")

    def __str__(self):
        return f"Name: {self.name}\nPhone: {self.phone_no}\n"

Cont1 = Contact("Anita", 9876543210)
print(Cont1)
```

## 4. Product Price Tag Generator 
**Problem Statement**: Create a class that stores product name and price and prints a formatted price tag. 

**Sample Input**<br>
Product Name: Headphones <br>
Price: 2499 <br>

**Expected Output**<br>
Product: Headphones <br>
Price: ₹2499<br>

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"Product: {self.name}\nPrice: ₹{self.price}\n"

pd1 = Product("Headphones", 2499)
print(pd1)
```

## 5. Movie Rating Display System 
**Problem Statement**: Create a class that stores movie name and rating and displays movie details. 

**Sample Input**:<br>
Movie Name: Inception<br> 
Rating: 4.8 <br>
 

**Expected Output**:<br>
Movie: Inception <br>
Rating: 4.8 / 5<br>

```python
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def __str__(self):
        return f"Movie: {self.name}\nRating: {self.rating} / 5\n"

movie1 = Movie("Inception", 4.8)
movie2 = Movie("Interstellar", 4.35)

print(movie1)
print(movie2)
```

## 6. Delivery Address Manager 
**Problem Statement**: Create a class that stores customer name and delivery address and prints delivery details.

**Sample Input**:<br>
Customer Name: Suman <br>
Address: Hyderabad <br>

**Expected Output**:<br>
Delivery Details <br>
Customer: Suman <br>
Address: Hyderabad<br>

```python
class Customer:
    def __init__(self, name, address):
        self.name = name
        self.address = address 
    
    def __str__(self):
        return f"Delivery Details\nName: {self.name}\nAddress: {self.address}\n"

cus1 = Customer("Suman", "Hyderbad")
cus2 = Customer("Ali", "Delhi")

print(cus1)
print(cus2)
```
