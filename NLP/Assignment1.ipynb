### **Task 1: Conceptual Understanding**

**1. What is the difference between "Love" and "love" in NLP?**
    <div style="margin: 10px 10px 0 2ch;">
    <b>Ans.</b> Computers are literal. To the algorithm, "Love" and "love" are totally different because their Underlying codes (binary/ASCII) don't match. If we don't fix this, the model treats them as two separate ideas, which thins out our data. We lowercase everything so the model realizes they are the same word, though we sometimes keep the caps if we are looking for specific names or brands.
    </div>

**2. What happens if stopwords are not removed?**
    <div style="margin: 10px 10px 0 2ch;">
    <b>Ans.</b> Stopwords like 'the', 'is' and 'and' are everywhere but don't actually tell us what a sentence is about. If we leave them in, they act like background noise that can drown out the important keywords. Plus, it makes the datasets way bulkier, which slows down training for no real reason.
    </div>

**3. Provided two real-world scenarios where removing stopwords can be harmful.**
    <div style="margin: 10px 10px 0 2ch;">
    <b>Ans.</b>
- **Sentiment Analysis:** If you have the sentence "I do not like this," and you delete 'not' the model thinks you are happy. That's a huge fail for catching negative statements.
- **Style/Author ID:** Everyone has a unique 'rhythm' in how they use small words. If you are trying to figure out who wrote a specific text, those stopwords are actually the 'fingerprints' you need to identity the author.
    </div>

**4. Explain the difference between stemming and lemmatization.**
    <div style="margin: 10px 10px 0 2ch;">
    <b>Ans.</b> Think of **Stemming** as a quick hack - it just chops off suffixes (like 'ing' or 'ed'). It is fast, but you often end up with non-words like 'studi'. <br><br>
    **Lemmatization** is the 'smarter' version. It looks at a dictionary to find the actual root word. It takes a bit longer because it needs to know if the word is a noun or a verb, but the result (like 'changing' --> 'change') actually make sense.
    </div>


### **Task 2: Build Advanced Preprocessing Function**
Implement the following function
```python
def preprocess_text(text):
    pass
```

**Required Features:**
1. Remove numbers
2. Remove extra spaces
3. Handle repeated characters
4. Convert text to lowercase
5. Remove very short tokens
6. Remove URLs and email-like patterns

````python
import re

def preprocess_text(text):
    # 1. Covert to lowercase
    text = text.lower()
    # print(f"Lowering case:- {text}")

    # 2. Remove URLs and Email-like patterns
    text = re.sub(r'https?://\S+|www\.\S+|\S+@\S+', '', text)
    # print(f'Remove URLs:- {text}')

    # 3. Handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    # print(f'Removed repeated characters:- {text}')

    # 4. Remove numbers
    text = re.sub(r'\d+', '', text)
    # print(f'Removed numbers:- {text}')

    # 5. Tokenize
    tokens = text.split()
    # print(f'After tokenization:- \n{tokens}')

    # 6. Filter tokens
    clear_tokens = []
    for t in tokens:
        t = re.sub(r'[^\w\s]', '', t)

        if len(t) > 2 or t in ['no','not']:
            if t:
                clear_tokens = clear_tokens + [t]

    clean_sentence = " ".join(clear_tokens)

    return clear_tokens, clean_sentence


sample_text = "Hey <rishi@example.com>, Visit https://google.com for 100% FREE stuuuuuuffff!!! No way!"
tokens, sentence = preprocess_text(sample_text)
print(f'Clear_token:- {tokens}\nClean_sentence:- {sentence}')

````

### **Task 4: Token Analytics**
For each processed sentence, complete
1. Total number of tokens.
2. Number of unique tokens
3. Average token length

````python
results = []
for s in test_sentences:
    tokens, cleaned = preprocess_text(s)
    
    # Basic math for analytics
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    # Avoid division by zero if a sentence is empty after cleaning
    avg_length = sum(len(t) for t in tokens) / total_tokens if total_tokens > 0 else 0
    
    results.append({
        "original": s,
        "tokens": tokens,
        "count": total_tokens,
        "unique": unique_tokens,
        "avg_len": round(avg_length, 2)
    })

# Displaying the results clearly
print(f"{'Original Sentence':<35} | {'Tokens':<5} | {'Unique':<6} | {'Avg Len'}")
print("-" * 70)
for r in results:
    print(f"{r['original'][:34]:<35} | {r['count']:<5} | {r['unique']:<6} | {r['avg_len']}")
````

````python
from collections import Counter

# 1. Collect every token from every sentence we processed earlier
all_tokens = []
for s in test_sentences:
    tokens, _ = preprocess_text(s)
    all_tokens.extend(tokens)

# 2. Use Counter to do the heavy lifting
word_counts = Counter(all_tokens)

# 3. Identify Top 10 Most Frequent
top_10 = word_counts.most_common(10)

# 4. Identify Top 5 Least Frequent 
least_5 = word_counts.most_common()[:-6:-1]

print("--- Top 10 Most Frequent Words ---")
for word, count in top_10:
    print(f"{word}: {count}")

print("\n--- Top 5 Least Frequent Words ---")
for word, count in least_5:
    print(f"{word}: {count}")
````

### **Task 6: Build Full Pipeline**

```python
def full_pipeline(test_list):
```

**Expected Output**
```json
{
    "tokens": [...],
    "clean_sentences": [...]
}
```


````python
import json

def full_pipeline(text_list):
    all_tokens = []
    all_clean_sentences = []
    
    for text in text_list:
        # 1. Standardize Case
        text = text.lower()
        
        # 2. Scrub URLs and Emails
        text = re.sub(r'https?://\S+|www\.\S+|<\S+@\S+>', '', text)
        
        # 3. Fix Repeated Chars (e.g., "noooo" -> "noo")
        text = re.sub(r'(.)\1{2,}', r'\1\1', text)
        
        # 4. Strip Numbers and Punctuation
        text = re.sub(r'\d+', '', text)
        text = re.sub(r'[^\w\s]', '', text)
        
        # 5. Tokenize and Filter
        tokens = text.split()
        clean_tokens = [t for t in tokens if len(t) > 2 or t in ['no', 'not']]
        
        # 6. Store Results
        all_tokens.extend(clean_tokens)
        all_clean_sentences.append(" ".join(clean_tokens))
        
    # Return the final structured format requested
    return {
        "tokens": all_tokens,
        "clean_sentences": all_clean_sentences
    }

# Running it on your sample data
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

output = full_pipeline(sample_inputs)


print(json.dumps(output, indent=4))
````

### **Task 7: Error Handling**
Ensure your solution handles:
- Empty string
- Only emojis
- Only numbers

````python
def preprocess_text_robust(text):
    # 1. Handle non-string inputs or empty strings
    if not isinstance(text, str) or not text.strip():
        return [], ""

    # 2. Case Normalization
    text = text.lower()
    
    # 3. URL/Email Removal
    text = re.sub(r'https?://\S+|www\.\S+|<\S+@\S+>', '', text)
    
    # 4. Repeated Character Reduction
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # 5. Number and Symbol Removal
    # This naturally handles "Only Numbers" or "Only Emojis/Symbols"
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    
    # 6. Tokenization
    tokens = text.split()
    
    # 7. Filtering (Length > 2 or "no"/"not")
    clean_tokens = [t for t in tokens if len(t) > 2 or t in ['no', 'not']]
    
    # 8. Final Check: Did we lose everything?
    if not clean_tokens:
        return [], ""
        
    clean_sentence = " ".join(clean_tokens)
    return clean_tokens, clean_sentence

# --- TESTING THE "DANGEROUS" INPUTS ---
edge_cases = [
    "",                # Empty string
    "   ",             # Only spaces
    "🔥🔥🔥😍😍",      # Only emojis
    "12345 678 90",    # Only numbers
    None,               # Not a string at all
    "This is normal sentence" # Normal sentence
]

print(f"{'Input Type':<15} | {'Tokens':<30} | {'Clean Sentence'}")
print("-" * 80)

for item in edge_cases:
    t, s = preprocess_text_robust(item)
    label = str(type(item)).split("'")[1]
    print(f"{label:<15} | {str(t):<30} | '{s}'")
````
